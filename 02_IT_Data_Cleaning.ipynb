```python
# Import all libraries needed
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder

print("All libraries imported successfully!")
```

```python
before = df.shape[0]
df = df.drop_duplicates()
after = df.shape[0]
print(f"Removed {before - after} duplicate rows")
print(f"Rows remaining: {after}")
```

```python
df['Date']        = pd.to_datetime(df['Date'])
df['Year']        = df['Date'].dt.year
df['Month']       = df['Date'].dt.month
df['Day_of_Week'] = df['Date'].dt.dayofweek

print("Date column fixed!")
print(df[['Date','Year','Month','Day_of_Week']].head())
```

```python
# Fill number columns with median
num_cols = ['CPU_Usage_pct', 'RAM_Usage_pct', 'Disk_Usage_pct',
            'Temperature_C', 'Network_Latency_ms',
            'Power_Consumption_W', 'Battery_Backup_pct',
            'Fan_Speed_RPM', 'Voltage_V']

for col in num_cols:
    missing = df[col].isnull().sum()
    if missing > 0:
        df[col] = df[col].fillna(df[col].median())
        print(f"Filled {missing} missing values in '{col}'")

# Fill text columns
df['Operating_System']  = df['Operating_System'].fillna('Unknown')
df['Failure_Component'] = df['Failure_Component'].fillna('None')
df['Failure_Type']      = df['Failure_Type'].fillna('None')

print("\nMissing values remaining:")
print(df.isnull().sum()[df.isnull().sum() > 0])
print("Done filling missing values!")
```

```python
ranges = {
    'CPU_Usage_pct'      : (0,   100),
    'RAM_Usage_pct'      : (0,   100),
    'Disk_Usage_pct'     : (0,   100),
    'Temperature_C'      : (10,  100),
    'Network_Latency_ms' : (0.1, 500),
    'Power_Consumption_W': (5,   2000),
    'Battery_Backup_pct' : (0,   100),
    'Fan_Speed_RPM'      : (100, 6000),
    'Voltage_V'          : (180, 260),
}

for col, (lo, hi) in ranges.items():
    outliers = ((df[col] < lo) | (df[col] > hi)).sum()
    if outliers > 0:
        df[col] = df[col].clip(lo, hi)
        print(f"Fixed {outliers} outliers in '{col}'")

print("Outlier fixing done!")
```

```python
for col in ['Device_Type', 'Location', 'Shift',
            'Maintenance_Priority', 'Operating_System']:
    le = LabelEncoder()
    df[col + '_enc'] = le.fit_transform(df[col])
    print(f"'{col}' encoded successfully")
```

```python
df['CPU_RAM_Avg']         = ((df['CPU_Usage_pct'] + df['RAM_Usage_pct']) / 2).round(2)
df['Temp_Fan_Ratio']      = (df['Temperature_C'] / df['Fan_Speed_RPM'] * 1000).round(4)
df['Resource_Pressure']   = (df['CPU_Usage_pct']*0.4 +
                              df['RAM_Usage_pct']*0.3 +
                              df['Disk_Usage_pct']*0.3).round(2)
df['Overdue_Maintenance'] = (df['Last_Maintenance_Days'] > 180).astype(int)
df['Critical_Temp']       = (df['Temperature_C'] > 70).astype(int)
df['High_Error']          = (df['Error_Logs_Count'] > 10).astype(int)
df['Age_Maint_Score']     = (df['Device_Age_Years'] *
                              df['Last_Maintenance_Days'] / 100).round(2)

print("New features created!")
print("New columns:", ['CPU_RAM_Avg','Temp_Fan_Ratio','Resource_Pressure',
                       'Overdue_Maintenance','Critical_Temp',
                       'High_Error','Age_Maint_Score'])
```
